```python
# Personal Expense Tracker

import csv
import os

expenses = []
budget = 0.0

FILE_NAME = "expenses.csv"

print("Personal Expense Tracker")
```

```python
def add_expense():
    date = input("Enter date (YYYY-MM-DD): ").strip()
    category = input("Enter category: ").strip()
    
    try:
        amount = float(input("Enter amount: "))
    except ValueError:
        print("Invalid amount. Please enter a number.")
        return
    
    description = input("Enter description: ").strip()

    # Validate required fields
    if not date or not category or not description:
        print("All fields are required. Expense not added.")
        return

    if amount < 0:
        print("Amount cannot be negative.")
        return

    expense = {
        "date": date,
        "category": category,
        "amount": amount,
        "description": description
    }

    expenses.append(expense)

    print("Expense added successfully.")
```

```python
def view_expenses():
    if not expenses:
        print("No expenses recorded.")
        return

    print("\n--- All Expenses ---")

    for expense in expenses:
        # Validate required details before displaying
        if not all(
            key in expense and expense[key] not in ("", None)
            for key in ["date", "category", "amount", "description"]
        ):
            print("Incomplete expense found. Skipping entry.")
            continue

        print(f"Date        : {expense['date']}")
        print(f"Category    : {expense['category']}")
        print(f"Amount      : ₹{expense['amount']:.2f}")
        print(f"Description : {expense['description']}")
        print("-" * 35)
```

```python
def track_budget():
    global budget

    try:
        budget = float(input("Enter your monthly budget: "))
    except ValueError:
        print("Invalid budget. Please enter a number.")
        return

    if budget < 0:
        print("Budget cannot be negative.")
        return

    total_expenses = sum(expense["amount"] for expense in expenses)

    print(f"\nMonthly Budget : ₹{budget:.2f}")
    print(f"Total Expenses : ₹{total_expenses:.2f}")

    if total_expenses > budget:
        exceeded = total_expenses - budget
        print(f"You have exceeded your budget by ₹{exceeded:.2f}!")
    else:
        remaining = budget - total_expenses
        print(f"You have ₹{remaining:.2f} left for the month.")
```

```python
def save_expenses():
    with open(FILE_NAME, "w", newline="", encoding="utf-8") as file:
        fieldnames = ["date", "category", "amount", "description"]

        writer = csv.DictWriter(file, fieldnames=fieldnames)

        writer.writeheader()
        writer.writerows(expenses)

    print(f"Expenses saved successfully to {FILE_NAME}.")
```

```python
def load_expenses():
    if not os.path.exists(FILE_NAME):
        print("No previous expense file found. Starting with an empty list.")
        return

    expenses.clear()

    with open(FILE_NAME, "r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        for row in reader:
            try:
                expense = {
                    "date": row["date"],
                    "category": row["category"],
                    "amount": float(row["amount"]),
                    "description": row["description"]
                }

                expenses.append(expense)

            except (KeyError, ValueError):
                print("Invalid expense found in file. Skipping entry.")

    print(f"{len(expenses)} expense(s) loaded successfully.")
```

```python
def display_menu():
    print("\n========== Personal Expense Tracker ==========")
    print("1. Add Expense")
    print("2. View Expenses")
    print("3. Track Budget")
    print("4. Save Expenses")
    print("5. Exit")
    print("==============================================")
```

```python
# Load previously saved expenses when the program starts
load_expenses()

while True:
    display_menu()

    choice = input("Enter your choice (1-5): ").strip()

    if choice == "1":
        add_expense()

    elif choice == "2":
        view_expenses()

    elif choice == "3":
        track_budget()

    elif choice == "4":
        save_expenses()

    elif choice == "5":
        save_expenses()
        print("Thank you for using Personal Expense Tracker.")
        print("Program exited successfully.")
        break

    else:
        print("Invalid choice. Please select a number between 1 and 5.")
```
